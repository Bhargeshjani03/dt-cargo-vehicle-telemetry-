```python
from pathlib import Path
import pandas as pd
import numpy as np

SPEED_ROOT = PROJECT_ROOT / "data" / "raw" / "speed"

raw_results = []

for _, row in sample.iterrows():

    track_id = int(row["track_id"])
    vehicle_id = int(row["vehicle_id"])

    speed_path = SPEED_ROOT / str(vehicle_id) / f"{track_id}.csv"

    if not speed_path.exists():
        raw_results.append({
            "track_id": track_id,
            "vehicle_id": vehicle_id,
            "file_exists": False
        })
        continue

    speed_df = pd.read_csv(speed_path)

    # Keep only valid speed observations
    speed_valid = speed_df["speed"].dropna()

    if len(speed_valid) > 0:
        raw_mean = speed_valid.mean()
        raw_max = speed_valid.max()
    else:
        raw_mean = np.nan
        raw_max = np.nan

    # Telemetry time span
    if len(speed_df) > 1:
        epoch = pd.to_numeric(speed_df["epoch"], errors="coerce").dropna()
        telemetry_span = epoch.max() - epoch.min()
    else:
        telemetry_span = np.nan

    raw_results.append({
        "track_id": track_id,
        "vehicle_id": vehicle_id,
        "file_exists": True,
        "raw_observations": len(speed_df),
        "raw_mean_speed": raw_mean,
        "raw_max_speed": raw_max,
        "telemetry_span": telemetry_span,

        # Track-level values
        "track_avg_speed": row["avg_speed"],
        "track_max_speed": row["max_speed"],
        "distance_m": row["distance"],
        "duration_s": row["duration_seconds"],

        # Speed calculated from metadata
        "distance_duration_speed": (
            row["distance"] / row["duration_seconds"]
            if row["duration_seconds"] > 0
            else np.nan
        )
    })

raw_check = pd.DataFrame(raw_results)

raw_check
```

```python
import pandas as pd
import numpy as np
from pathlib import Path

SPEED_ROOT = PROJECT_ROOT / "data" / "raw" / "speed"
OUTPUT_PATH = PROJECT_ROOT / "data" / "interim" / "speed_features.csv"

speed_records = []

for vehicle_dir in SPEED_ROOT.iterdir():

    if not vehicle_dir.is_dir():
        continue

    vehicle_id = int(vehicle_dir.name)

    for speed_file in vehicle_dir.glob("*.csv"):

        try:
            df = pd.read_csv(speed_file)

            if df.empty:
                speed_records.append({
                    "vehicle_id": vehicle_id,
                    "track_id": int(speed_file.stem),
                    "raw_observations": 0,
                    "telemetry_span_seconds": np.nan,
                    "raw_speed_mean": np.nan,
                    "raw_speed_std": np.nan,
                    "raw_speed_min": np.nan,
                    "raw_speed_max": np.nan,
                    "raw_hdop_mean": np.nan,
                    "raw_hdop_median": np.nan,
                    "raw_hdop_max": np.nan,
                    "missing_speed_count": 0,
                    "missing_hdop_count": 0
                })
                continue

            df["epoch"] = pd.to_numeric(
                df["epoch"],
                errors="coerce"
            )

            df["speed"] = pd.to_numeric(
                df["speed"],
                errors="coerce"
            )

            df["hdop"] = pd.to_numeric(
                df["hdop"],
                errors="coerce"
            )

            valid_epoch = df["epoch"].dropna()

            if len(valid_epoch) >= 2:
                telemetry_span = (
                    valid_epoch.max() - valid_epoch.min()
                )
            else:
                telemetry_span = np.nan

            speed = df["speed"]
            hdop = df["hdop"]

            speed_records.append({
                "vehicle_id": vehicle_id,
                "track_id": int(speed_file.stem),

                "raw_observations": len(df),

                "telemetry_span_seconds": telemetry_span,

                "raw_speed_mean": speed.mean(),
                "raw_speed_std": speed.std(),
                "raw_speed_min": speed.min(),
                "raw_speed_max": speed.max(),

                "raw_hdop_mean": hdop.mean(),
                "raw_hdop_median": hdop.median(),
                "raw_hdop_max": hdop.max(),

                "missing_speed_count": speed.isna().sum(),
                "missing_hdop_count": hdop.isna().sum()
            })

        except Exception as e:

            print(
                f"Error processing: {speed_file}"
            )

speed_features = pd.DataFrame(speed_records)

speed_features.to_csv(
    OUTPUT_PATH,
    index=False
)

print("Finished.")
print(f"Rows: {len(speed_features):,}")
print(f"Saved to: {OUTPUT_PATH}")
```

```python
import pandas as pd
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent

TRACKS_PATH = (
    PROJECT_ROOT / "data" / "raw" / "tracks" / "tracks.csv"
)

SPEED_FEATURES_PATH = (
    PROJECT_ROOT / "data" / "interim" / "speed_features.csv"
)

tracks = pd.read_csv(TRACKS_PATH)

if not SPEED_FEATURES_PATH.exists():
    raise FileNotFoundError(
        f"Speed features file not found: {SPEED_FEATURES_PATH}"
    )

speed_features = pd.read_csv(SPEED_FEATURES_PATH)

print("Tracks shape:", tracks.shape)
print("Speed features shape:", speed_features.shape)

print(
    "Unique speed-feature track IDs:",
    speed_features["track_id"].nunique()
)

print(
    "Duplicate track IDs:",
    speed_features["track_id"].duplicated().sum()
)

print(
    "Missing track IDs:",
    speed_features["track_id"].isna().sum()
)

coverage = tracks[["track_id"]].merge(
    speed_features[["track_id"]],
    on="track_id",
    how="outer",
    indicator=True
)

print("\nTrack ID coverage:")
print(coverage["_merge"].value_counts())
```
